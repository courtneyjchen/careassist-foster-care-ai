```python
#make sure we're counting by unique ID (STFCID = State Foster Care Identifier )
df_unique = data.drop_duplicates(subset='StFCID')
```

```python
#MOST COMMON REMOVAL REASON VISUALIZATION

removal_cols = [
    'NEGLECT',
    'PHYABUSE',
    'SEXABUSE',
    'DAPARENT',
    'AAPARENT',
    'DACHILD',
    'AACHILD',
    'PRTSJAIL',
    'CHBEHPRB',
    'HOUSING',
    'NOCOPE',
    'ABANDMNT',
    'RELINQSH',
    'PRTSDIED',
    'CHILDIS'
]

removal_counts = (
    data[removal_cols]
    .sum()
    .sort_values(ascending=False)
)

removal_pct = removal_counts / len(data) * 100

```

```python
plt.figure()
removal_counts.plot(kind="barh")
plt.gca().invert_yaxis()
plt.xlabel("Number of Children")
plt.title("Most Common Reasons for Removal")
plt.show()
```

```python
#DISTRIBUTION OF HOW LONG CHILDREN ARE IN THE SYSTEM
# using NumPlep = Number of Placements -- make sure to convert to INT

# Make sure the column is integer and handle missing NA
df_plot = df_unique.dropna(subset=['NUMPLEP']).copy() #drop children with unknown NUMPLEP
df_plot['NUMPLEP'] = df_plot['NUMPLEP'].astype(int)
max_display = df_plot['NUMPLEP'].max()
df_plot['NUMPLEP_trim'] = df_plot['NUMPLEP'].clip(upper=max_display)

# Plot histogram
plt.figure(figsize=(8,5))

plt.hist(
    df_plot['NUMPLEP_trim'],
    bins=range(0, max_display + 2),   # +2 is critical
    edgecolor='black',
    align='left'
)

plt.xlabel('Number of Placement Changes')
plt.ylabel('Number of Children')
plt.title(f'Distribution of Placement Changes (NUMPLEP, {max_display}+ grouped)')

# Control x-ticks to prevent overlap
plt.xticks(range(0, max_display + 1))

plt.tight_layout()
plt.show()
```

```python
#What is the rate of children getting adopted vs. ageing out of the system?

# Keep latest record per child (by discharge date or FY)
df_child = (
    data.sort_values('DoDFCDt')
      .drop_duplicates(subset='StFCID', keep='last')
)

# Outcome flags
df_child['Adopted'] = df_child['DISREASN'].eq(3) #3 = ADOPTION
df_child['AgedOut_flag'] = df_child['AgedOut'] == 1
```

```python
total_children = len(df_child)

counts = {
    'Adopted': df_child['Adopted'].sum(),
    'Aged Out': df_child['AgedOut_flag'].sum()
}
```

```python
labels = list(counts.keys())
values = list(counts.values())

colors = ['#4CAF50', '#F44336']  # green = adoption, red = aged out

plt.figure(figsize=(6,4))
plt.bar(labels, values, color=colors)
plt.ylabel('Number of Children')
plt.title('Number of Children Adopted vs. Aged Out')

for i, v in enumerate(values):
    plt.text(i, v, f"{v:,}", ha='center', va='bottom')

plt.tight_layout()
plt.show()
```

```python
discharge_labels = {
    0: 'Not Applicable',
    1: 'Reunified with Parent / Primary Caretaker',
    2: 'Living with Other Relative(s)',
    3: 'Adoption',
    4: 'Emancipation',
    5: 'Guardianship',
    6: 'Transfer to Another Agency',
    7: 'Runaway',
    8: 'Death of Child'
}
```

```python
df_child = (
    data.sort_values('DoDFCDt')
      .drop_duplicates(subset='StFCID', keep='last')
      .copy()
)

df_child['DischargeReason'] = (
    df_child['DISREASN']
      .map(discharge_labels)
      .fillna('Unknown')
)

discharge_counts = (
    df_child['DischargeReason']
      .value_counts()
      .sort_values(ascending=True)
)

plt.figure(figsize=(8,5))
plt.barh(discharge_counts.index, discharge_counts.values)
plt.xlabel('Number of Children')
plt.title('Number of Children by Discharge Reason')

for i, v in enumerate(discharge_counts.values):
    plt.text(v, i, f"{v:,}", va='center')

plt.tight_layout()
plt.show()

```

### Selected Metrics & Definitions

| Variable | Definition | Why It’s Important |
|--------|------------|---------------------------------------------|
| **AgeAtLatRem** | Age of the child at the most recent removal from home (entry or re-entry into foster care). | One of the strongest predictors of length of stay, permanency outcomes, and aging out. Anchors analysis at the moment of system intervention and is essential for stratifying all downstream outcomes. |
| **Sex** | Biological sex of the child, as reported. | Enables early detection of systematic differences in placement instability, behavioral-related removals, and aging-out risk. Serves as a basic but essential stratifier. |
| **RaceEthn** | Derived, mutually exclusive race and ethnicity classification combining race and Hispanic/Latino ethnicity. | Central for identifying racial and ethnic disparities in entry, experience, and outcomes. Using the derived variable avoids double counting and ensures consistency with federal reporting standards. |
| **Neglect** | Indicator that neglect was a reported reason for removal from home. | Typically the most common removal reason. Neglect-driven cases often have longer stays and different reunification dynamics, making it critical for understanding system drivers. |
| **DAParent** | Indicator that parental drug abuse was a reported reason for removal. | Strongly associated with repeat removals, longer time to permanency, and higher likelihood of TPR. Represents a distinct intervention and service pathway beyond neglect alone. |
| **NumPlep** | Number of placement settings during the current foster care episode (placement changes). | Primary measure of placement instability. High values are linked to trauma, behavioral escalation, and poorer outcomes, making this a key system-performance metric. |
| **LifeLOS** | Total cumulative number of days the child has spent in foster care across all episodes. | Captures total system exposure rather than a single episode. Highly skewed and ideal for exploratory distribution analysis; closely tied to long-term outcomes. |
| **CurPlSet** | Current placement setting (e.g., foster family home, kinship care, group home, institution). | Reflects level of care, system capacity, and policy choices. Strongly associated with stability, length of stay, and age-specific placement patterns. |
| **DISREASN** | Reason the child was discharged from foster care (e.g., reunification, adoption, guardianship, aging out). | Defines case outcomes and serves as a natural endpoint for analysis. Essential for distinguishing system success versus failure. |
| **AgedOut** | Indicator that the youth exited foster care due to reaching the age limit for eligibility. | Often considered a high-risk or negative outcome. Closely tied to late entry, long stays, and placement instability, making it a critical outcome metric. |
